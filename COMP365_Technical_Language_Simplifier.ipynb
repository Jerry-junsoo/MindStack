
## 9. Test Qwen3-4B with a Small Batch of Passages

After evaluating the initial Qwen2.5-1.5B-Instruct model, the team introduced Qwen3-4B to improve the quality of technical simplifications.

The first three technical passages were used as an initial test of Qwen3-4B. The model was instructed to use common words and clear sentences, preserve the original technical meaning, and avoid adding unsupported facts, examples, or comparisons.

The generated passages were examined to determine whether the new model could retain important technical details while making the text easier to understand.

After the initial test, Qwen3-4B was used to simplify all 30 passages in the dataset.

**Expected result:** Three sample passages are simplified and displayed for review before processing the complete dataset.


```python

# Simplify the first three passages using Qwen3-4B

import torch

batch_df_v2 = df.head(3).copy()

def simplify_text_v2(text):
    prompt = f"""Simplify the following technical passage for a middle-school reader.

Accuracy is more important than making the passage sound interesting.

STRICT RULES:
1. Preserve the original meaning and all important technical information.
2. Keep every number, date, time period, measurement, and technical name exactly as stated.
3. Never change, estimate, round, or convert numerical information.
4. Do not invent facts, examples, comparisons, metaphors, stories, or explanations.
5. Do not use analogies such as playgrounds, toys, or everyday objects.
6. Do not claim that something is faster, safer, easier, better, or more reliable unless the original explicitly says so.
7. Keep technical terms when replacing them would change their meaning.
8. Use common words and short, clear sentences.
9. Do not remove important details to make the passage shorter.
10. If a technical idea is difficult to simplify accurately, keep the original technical wording rather than guessing.
11. Return only the simplified passage.
12. Do not replace technical terms with explanations that have a different meaning. For example, keep "authenticates" and "authorizes" distinct.
13. Preserve specialized technical terms exactly when simplifying them could change their meaning. In particular, do not replace "authenticate" or "authorize" with general explanations.

Original passage:
{text}"""

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    prompt_text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
        enable_thinking=False
    )

    inputs = tokenizer(
        prompt_text,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=512,
            do_sample=False
        )

    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

    simplified_text = tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    )

    return simplified_text.strip()


batch_df_v2["Simplified_Text"] = ""

for index, row in batch_df_v2.iterrows():
    print(f"\n{'=' * 70}")
    print(f"Passage: {row['Passage_ID']} | Category: {row['Category']}")
    print("\nORIGINAL:")
    print(row["Original_Text"])

    simplified = simplify_text_v2(row["Original_Text"])
    batch_df_v2.at[index, "Simplified_Text"] = simplified

    print("\nQWEN3-4B VERSION 2:")
    print(simplified)

print("\nQwen3-4B small-batch test completed.")
```

```python
# Prepare all 30 passages for Version 2

batch_df_v2 = df.copy()

batch_df_v2["Simplified_Text"] = ""

print("Total passages for Version 2:", len(batch_df_v2))
print("\nPassages by category:")
print(batch_df_v2["Category"].value_counts())
```


## 12. Measure Readability

Readability scores are calculated for the original passages and the simplified passages produced by both language models. This evaluation helps determine whether the models made the technical content easier to read.

The **Flesch Reading Ease** score estimates how easy a passage is to read. Higher scores generally indicate easier text. The **Flesch-Kincaid Grade Level** estimates the U.S. school grade level needed to understand the text. Lower grade-level scores generally indicate easier text.

The results for the original passages, Version 1 (Qwen2.5-1.5B-Instruct), and Version 2 (Qwen3-4B) are compared to identify changes in readability.

These scores measure reading difficulty, not whether the simplified text preserves the original meaning or factual accuracy. Meaning preservation and factual accuracy are evaluated separately through human review.


```python
# Install the readability library
!pip -q install textstat
```

```python
# Calculate readability scores for original and simplified passages

import textstat
import pandas as pd

readability_df = results_df.copy()

readability_df["Original_Flesch_Reading_Ease"] = (
    readability_df["Original_Text"].apply(
        textstat.flesch_reading_ease
    )
)

readability_df["Simplified_Flesch_Reading_Ease"] = (
    readability_df["Simplified_Text"].apply(
        textstat.flesch_reading_ease
    )
)

readability_df["Original_Grade_Level"] = (
    readability_df["Original_Text"].apply(
        textstat.flesch_kincaid_grade
    )
)

readability_df["Simplified_Grade_Level"] = (
    readability_df["Simplified_Text"].apply(
        textstat.flesch_kincaid_grade
    )
)

readability_df["Reading_Ease_Change"] = (
    readability_df["Simplified_Flesch_Reading_Ease"]
    - readability_df["Original_Flesch_Reading_Ease"]
)

readability_df["Grade_Level_Change"] = (
    readability_df["Simplified_Grade_Level"]
    - readability_df["Original_Grade_Level"]
)

print("Readability scores calculated for all passages.")
print("Total passages:", len(readability_df))

display(
    readability_df[
        [
            "Passage_ID",
            "Original_Flesch_Reading_Ease",
            "Simplified_Flesch_Reading_Ease",
            "Original_Grade_Level",
            "Simplified_Grade_Level",
        ]
    ].round(2)
)
```


## Jordyn Walsh-Hill — Checkpoint 3 Human Quality Evaluation

This section evaluates whether the simplified passages preserve the original technical meaning and remain factually accurate.

The review uses two criteria:

- **Meaning Preserved:** The simplified passage retains the essential technical information, important details, and original meaning.
- **Factually Accurate:** The simplified passage does not introduce unsupported claims, incorrect information, or changes to technical facts.

The purpose of this evaluation is to identify weaknesses in the AI simplification process and determine what improvements are needed before the final presentation.


```python

# Jordyn Walsh-Hill
# Checkpoint 3: Human Quality Evaluation

import pandas as pd

#Standardize review responses
review_df["Meaning_Preserved"] = (
    review_df["Meaning_Preserved"].astype(str).str.strip().str.lower()
)

review_df["Factually_Accurate"] = (
    review_df["Factually_Accurate"].astype(str).str.strip().str.lower()
)

# Count completed human evaluations
total = len(review_df)

meaning_yes = (
    review_df["Meaning_Preserved"] == "yes"
).sum()

accuracy_yes = (
    review_df["Factually_Accurate"] == "yes"
).sum()

both_yes = (
    (review_df["Meaning_Preserved"] == "yes") &
    (review_df["Factually_Accurate"] == "yes")
).sum()

print("JORDYN WALSH-HILL: HUMAN QUALITY EVALUATION")
print("=" * 50)

print("Total passages reviewed:", total)

print(
    f"Meaning preserved: {meaning_yes}/{total} "
    f"({meaning_yes/total*100:.1f}%)"
)

print(
    f"Factually accurate: {accuracy_yes}/{total} "
    f"({accuracy_yes/total*100:.1f}%)"
)

print(
    f"Both criteria satisfied: {both_yes}/{total} "
    f"({both_yes/total*100:.1f}%)"
)
```

```python

#Identify passages that failed one or both quality criteria
failed_passages = review_df[
    (review_df["Meaning_Preserved"] != "yes") |
    (review_df["Factually_Accurate"] != "yes")
].copy()

print("PASSAGES NEEDING QUALITY IMPROVEMENT")
print("=" * 50)
display(
    failed_passages[
        [
            "Passage_ID",
            "Meaning_Preserved",
            "Factually_Accurate",
            "Review_Notes"
        ]
    ]
)

print("\nTotal passages needing improvement:", len(failed_passages))
```
